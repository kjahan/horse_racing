```python
import pandas as pd
```

```python
# specify the value of each column

df = pd.DataFrame({"a" : [4, 5, 6],
                   "b" : [7, 8, 9],
                   "c" : [10, 11, 12]},
                  index = [1, 2, 3])
```

```python
df.head()
```

```python
# Specify values for each row.
df = pd.DataFrame([[4, 7, 10],
                   [5, 8, 11],
                   [6, 9, 12]],
                  index=[1, 2, 3],
                  columns=['a', 'b', 'c'])
```

```python
df.shape
```

```python
df.head()
```

```python
df.iloc[0]
```

```python
df.iloc[:, [1]]
```

```python
df.iloc[:, [1,2]]
```

```python
df.head()
```

```python
df.loc[df['a'] > 10]
```

```python
df.loc[df['a'] > 5]['a']
```

```python
# Access single value by index
df.iat[1, 2]
```

```python
# Access single value by label
df.at[1, 'a'] 
```

```python
df['a'].nunique()
# of distinct values in a column.
```

```python
df.describe()
```

```python
df.sum()
```

```python
# Drop rows with any column having NA/null data
df.dropna()
```

```python
# Replace all NA/null data with value.
df.fillna(0)
```
